# Exercise 33

Suppose \(A\) is an $ m \times n $ matrix and there exist $ n \times m $ matrices \(C\) and \(D\) such that:

$$
CA=I_n
$$

and

$$
AD=I_m
$$

Prove that:

$$
m=n
$$

and

$$
C=D
$$

**Hint:** Think about the product:

$$
CAD
$$

Since $CA=I_n$, multiplying on the right by $D$ gives

$$
CAD=I_nD=D.
$$

Since $AD=I_m$, multiplying on the left by $C$ gives

$$
CAD=CI_m=C.
$$

Therefore,

$$
C=CAD=D.
$$

Thus $D=C$, so

$$
DA=CA=I_n.
$$

Together with $AD=I_m$, this shows that $D$ is both a left inverse and a right inverse of $A$. A matrix can have a two-sided inverse only when it is square. Hence,

$$
m=n.
$$

# Section 2.4 - Exercise 17

Suppose $A_{11}$ is invertible. Find $X$ and $Y$ such that

$$
\begin{bmatrix} A_{11} & A_{12} \\ A_{21} & A_{22} \end{bmatrix}
=
\begin{bmatrix} I & 0 \\ X & I \end{bmatrix}
\begin{bmatrix} A_{11} & 0 \\ 0 & S \end{bmatrix}
\begin{bmatrix} I & Y \\ 0 & I \end{bmatrix}.
\tag{7}
$$

where $S=A_{22}-A_{21}A_{11}^{-1}A_{12}$. The matrix $S$ is called the **Schur complement** of $A_{11}$. Likewise, if $A_{22}$ is invertible, the matrix $A_{11}-A_{12}A_{22}^{-1}A_{21}$ is called the Schur complement of $A_{22}$. Such expressions occur frequently in the theory of systems engineering, and elsewhere.

## Solution

Multiplying the three block matrices gives

$$
\begin{bmatrix} I&0\\X&I\end{bmatrix}
\begin{bmatrix} A_{11}&0\\0&S\end{bmatrix}
\begin{bmatrix} I&Y\\0&I\end{bmatrix}
=\begin{bmatrix}A_{11}&A_{11}Y\\XA_{11}&XA_{11}Y+S\end{bmatrix}.
$$

Comparing blocks with the matrix on the left of (7),

$$A_{11}Y=A_{12}\quad\Longrightarrow\quad\boxed{Y=A_{11}^{-1}A_{12}},$$

and

$$XA_{11}=A_{21}\quad\Longrightarrow\quad\boxed{X=A_{21}A_{11}^{-1}}.$$

The lower-right block is then

$$
XA_{11}Y+S=A_{21}A_{11}^{-1}A_{12}
+A_{22}-A_{21}A_{11}^{-1}A_{12}=A_{22},
$$

so these choices give the required factorization.


# Section 2.4 - Exercise 21

In the study of engineering control of physical systems, a standard set of differential equations is transformed by Laplace transforms into the following system of linear equations:

$$
\begin{bmatrix} A-sI_n & B \\ C & I_m \end{bmatrix}
\begin{bmatrix} \mathbf{x} \\ \mathbf{u} \end{bmatrix}
=
\begin{bmatrix} \mathbf{0} \\ \mathbf{y} \end{bmatrix}.
\tag{8}
$$

where $A$ is $n\times n$, $B$ is $n\times m$, $C$ is $m\times n$, and $s$ is a variable. The vector $\mathbf{u}$ in $\mathbb{R}^m$ is the "input" to the system, $\mathbf{y}$ in $\mathbb{R}^m$ is the "output," and $\mathbf{x}$ in $\mathbb{R}^n$ is the "state" vector. (Actually, the vectors $\mathbf{x}$, $\mathbf{u}$, and $\mathbf{y}$ are functions of $s$, but we suppress this fact because it does not affect the algebraic calculations in Exercises 21 and 22.)

The result is an equation of the form $W(s)\mathbf{u}=\mathbf{y}$, where $W(s)$ is a matrix that depends on $s$. $W(s)$ is called the *transfer function* of the system because it transforms the input $\mathbf{u}$ into the output $\mathbf{y}$. Find $W(s)$ and describe how it is related to the partitioned system matrix on the left side of (8). See Exercise 17.

## Solution

The top block equation is

$$ (A-sI_n)\mathbf{x}+B\mathbf{u}=\mathbf{0}. $$

Because $A-sI_n$ is invertible,

$$ \mathbf{x}=-(A-sI_n)^{-1}B\mathbf{u}. $$

$$ \left[I_m-C(A-sI_n)^{-1}B\right]\mathbf{u}=\mathbf{y}. $$

Therefore,

$$ \boxed{W(s)=I_m-C(A-sI_n)^{-1}B}. $$

This is the Schur complement of $A-sI_n$ in

$$\begin{bmatrix}A-sI_n&B\\C&I_m\end{bmatrix}.$$

# Section 2.4 - Exercise 29

Suppose memory or size restrictions prevent your matrix program from working with matrices having more than 32 rows and 32 columns, and suppose some project involves $50\times 50$ matrices $A$ and $B$. Describe the commands or operations of your matrix program that accomplish the following tasks.

**a.** Compute $A+B$.

**b.** Compute $AB$.

**c.** Solve $A\mathbf{x}=\mathbf{b}$ for some vector $\mathbf{b}$ in $\mathbb{R}^{50}$, assuming that $A$ can be partitioned into a $2\times 2$ block matrix $[A_{ij}]$, with $A_{11}$ an invertible $20\times 20$ matrix, $A_{22}$ an invertible $30\times 30$ matrix, and $A_{12}$ a zero matrix. **Hint:** Describe appropriate smaller systems to solve, without using any matrix inverses.

## Solution

For parts (a) and (b), store each $50\times50$ matrix as four $25\times25$ blocks:

$$
A=\begin{bmatrix}A_{11}&A_{12}\\A_{21}&A_{22}\end{bmatrix},\qquad
B=\begin{bmatrix}B_{11}&B_{12}\\B_{21}&B_{22}\end{bmatrix}.
$$

### (a) Addition

Add corresponding blocks. The function below returns four small blocks and never assembles a $50\times50$ array.


```python
def add_block_matrices(A_blocks, B_blocks): 
    A11, A12, A21, A22 = A_blocks  
    B11, B12, B21, B22 = B_blocks  
    C11 = A11 + B11  
    C12 = A12 + B12  
    C21 = A21 + B21  
    C22 = A22 + B22 
    return C11, C12, C21, C22  
```

### (b) Multiplication

Use the block row-column rule:

$$
AB=\begin{bmatrix}
A_{11}B_{11}+A_{12}B_{21}&A_{11}B_{12}+A_{12}B_{22}\\
A_{21}B_{11}+A_{22}B_{21}&A_{21}B_{12}+A_{22}B_{22}
\end{bmatrix}.
$$


```python
def multiply_block_matrices(A_blocks, B_blocks): 
    A11, A12, A21, A22 = A_blocks  
    B11, B12, B21, B22 = B_blocks  
    C11 = A11 @ B11 + A12 @ B21 
    C12 = A11 @ B12 + A12 @ B22 
    C21 = A21 @ B11 + A22 @ B21 
    C22 = A21 @ B12 + A22 @ B22 
    return C11, C12, C21, C22  
```

### (c) Solving the system

Use the partition in the question:

$$
A=\begin{bmatrix}A_{11}&0\\A_{21}&A_{22}\end{bmatrix},\quad
\mathbf{x}=\begin{bmatrix}\mathbf{x}_1\\\mathbf{x}_2\end{bmatrix},\quad
\mathbf{b}=\begin{bmatrix}\mathbf{b}_1\\\mathbf{b}_2\end{bmatrix},
$$

where the block sizes are 20 and 30. First solve

$$A_{11}\mathbf{x}_1=\mathbf{b}_1,$$

and then solve

$$A_{22}\mathbf{x}_2=\mathbf{b}_2-A_{21}\mathbf{x}_1.$$

This uses only a $20\times20$ solve and a $30\times30$ solve, with no matrix inverse.


```python
import numpy as np 
def solve_lower_block_system(A11, A21, A22, b1, b2): 
    x1 = np.linalg.solve(A11, b1) 
    c = b2 - A21 @ x1 
    x2 = np.linalg.solve(A22, c) 
    return x1, x2  
```

# Section 2.5 - Exercise 11

Find an LU factorization of the matrix (with $L$ unit lower triangular). Note that MATLAB will usually produce a permuted LU factorization because it uses partial pivoting for numerical accuracy.

$$
A=\begin{bmatrix} 3 & -6 & 3 \\ 6 & -7 & 2 \\ -1 & 7 & 0 \end{bmatrix}.
$$

## Solution

Use the row replacements

$$R_2\leftarrow R_2-2R_1,\qquad R_3\leftarrow R_3+\frac13R_1,\qquad R_3\leftarrow R_3-R_2.$$

The elimination multipliers are $2$, $-1/3$, and $1$. Therefore,

$$
\boxed{L=\begin{bmatrix}1&0&0\\2&1&0\\-1/3&1&1\end{bmatrix}},\qquad
\boxed{U=\begin{bmatrix}3&-6&3\\0&5&-4\\0&0&5\end{bmatrix}},
$$

and $A=LU$.


```python
import sympy as sp 
from IPython.display import display  
A = sp.Matrix([[3, -6, 3], [6, -7, 2], [-1, 7, 0]]) 
L = sp.Matrix([[1, 0, 0], [2, 1, 0], [sp.Rational(-1, 3), 1, 1]])  
U = sp.Matrix([[3, -6, 3], [0, 5, -4], [0, 0, 5]]) 
display(L * U)  
print(L * U == A) 
```

Let $A$ be a lower triangular $n\times n$ matrix with nonzero entries on the diagonal. Also, explain why the row operations that reduce $A$ to $I$ change $I$ into a lower triangular matrix.

## Solution

Because $A$ is lower triangular and every diagonal entry is nonzero, every diagonal position is a pivot. Using row replacements and row scaling, $A$ can be reduced to $I_n$. Therefore $A$ is invertible by the Invertible Matrix Theorem.

Apply the same operations to $[A\mid I_n]$. The required row replacements add multiples of earlier rows to later rows, and scaling changes only one row. Since

$$[A\mid I_n]\longrightarrow[I_n\mid A^{-1}],$$

we conclude that

$$\boxed{A\text{ is invertible and }A^{-1}\text{ is lower triangular}.}$$

# Section 2.7 - Exercise 10

Consider the following geometric 2D transformations: $D$, a dilation (in which $x$-coordinates and $y$-coordinates are scaled by the same factor); $R$, a rotation; and $T$, a translation.

Does $D$ commute with $R$? That is, is

$$
D(R(\mathbf{x})) = R(D(\mathbf{x}))
$$

for all $\mathbf{x}$ in $\mathbb{R}^2$? Does $D$ commute with $T$? Does $R$ commute with $T$?

## Solution

Let the dilation factor be $s$, let $Q$ be the rotation matrix, and let the translation vector be $\mathbf{a}$. Then

$$
D(\mathbf{x})=s\mathbf{x},\qquad
R(\mathbf{x})=Q\mathbf{x},\qquad
T(\mathbf{x})=\mathbf{x}+\mathbf{a}.
$$

For dilation and rotation,

$$
D(R(\mathbf{x}))=sQ\mathbf{x}=Q(s\mathbf{x})=R(D(\mathbf{x})).
$$

Therefore, **$D$ commutes with $R$**.

For dilation and translation,

$$
D(T(\mathbf{x}))=s\mathbf{x}+s\mathbf{a},
\qquad
T(D(\mathbf{x}))=s\mathbf{x}+\mathbf{a}.
$$

These are not equal in general, so **$D$ does not commute with $T$**. They are equal only in special cases such as $s=1$ or $\mathbf{a}=\mathbf{0}$.

For rotation and translation,

$$
R(T(\mathbf{x}))=Q\mathbf{x}+Q\mathbf{a},
\qquad
T(R(\mathbf{x}))=Q\mathbf{x}+\mathbf{a}.
$$

These are not equal in general, so **$R$ does not commute with $T$**. Equality requires $Q\mathbf{a}=\mathbf{a}$.

**Knowledge point and takeaway.** A dilation has matrix $sI$, which commutes with every rotation matrix. Translation is not a linear transformation on $\mathbb{R}^2$, so the order of operations usually matters. In a graphics pipeline, changing the order of rotation, scaling, and translation can move an object to a completely different position.

we obtain

$$
A_2A_1=
\begin{bmatrix}
\cos\phi & -\sin\phi & 0\\
\sin\phi & \cos\phi & 0\\
0 & 0 & 1
\end{bmatrix}.
$$

This is exactly the homogeneous-coordinate matrix for a counterclockwise rotation through angle $\phi$ in $\mathbb{R}^2$.

**Knowledge point and takeaway.** Composite transformations are represented by matrix products, read from right to left. This factorization is useful in computer graphics because a rotation can be implemented by simple shear-and-scale steps that may be efficient at the pixel level.

# Section 2.7 - Exercise 13

The usual transformations on homogeneous coordinates for 2D computer graphics involve $3\times 3$ matrices of the form

$$
\begin{bmatrix}
A & \mathbf{p} \\
\mathbf{0}^{T} & 1
\end{bmatrix},
$$

where $A$ is a $2\times 2$ matrix and $\mathbf{p}$ is in $\mathbb{R}^2$. Show that such a transformation amounts to a linear transformation on $\mathbb{R}^2$ followed by a translation.

*Hint:* Find an appropriate matrix factorization involving partitioned matrices.

## Solution

Factor the homogeneous-coordinate matrix as

$$
\begin{bmatrix}
A & \mathbf{p}\\
\mathbf{0}^{T} & 1
\end{bmatrix}
=
\begin{bmatrix}
I & \mathbf{p}\\
\mathbf{0}^{T} & 1
\end{bmatrix}
\begin{bmatrix}
A & \mathbf{0}\\
\mathbf{0}^{T} & 1
\end{bmatrix}.
$$

Applied to $\begin{bmatrix}\mathbf{x}\\1\end{bmatrix}$, the right matrix acts first:

$$
\begin{bmatrix}
A & \mathbf{0}\\
\mathbf{0}^{T} & 1
\end{bmatrix}
\begin{bmatrix}\mathbf{x}\\1\end{bmatrix}
=
\begin{bmatrix}A\mathbf{x}\\1\end{bmatrix}.
$$

The left matrix then translates by $\mathbf{p}$:

$$
\begin{bmatrix}
I & \mathbf{p}\\
\mathbf{0}^{T} & 1
\end{bmatrix}
\begin{bmatrix}A\mathbf{x}\\1\end{bmatrix}
=
\begin{bmatrix}A\mathbf{x}+\mathbf{p}\\1\end{bmatrix}.
$$

Thus the transformation is **the linear transformation $\mathbf{x}\mapsto A\mathbf{x}$ followed by translation by $\mathbf{p}$**.

**Knowledge point and takeaway.** Homogeneous coordinates convert an affine map $\mathbf{x}\mapsto A\mathbf{x}+\mathbf{p}$ into one matrix multiplication. This is why graphics systems can store a complete sequence of linear transformations and translations in a single matrix.

# Section 2.7 - Exercise 14

Show that the transformation in Exercise 7 is equivalent to a rotation about the origin followed by a translation by $\mathbf{p}$. Find $\mathbf{p}$.

## Solution

Exercise 7 rotates points through $60^\circ$ about the center

$$
\mathbf{c}=\begin{bmatrix}6\\8\end{bmatrix}.
$$

To rotate about $\mathbf{c}$, first translate by $-\mathbf{c}$, rotate about the origin, and then translate back by $\mathbf{c}$:

$$
\mathbf{x}\longmapsto
R(\mathbf{x}-\mathbf{c})+\mathbf{c}
=R\mathbf{x}+(\mathbf{c}-R\mathbf{c}).
$$

For $60^\circ$,

$$
R=
\begin{bmatrix}
\tfrac12 & -\tfrac{\sqrt3}{2}\\
\tfrac{\sqrt3}{2} & \tfrac12
\end{bmatrix},
\qquad
R\mathbf{c}=
\begin{bmatrix}
3-4\sqrt3\\
3\sqrt3+4
\end{bmatrix}.
$$

Therefore,

$$
\mathbf{p}=\mathbf{c}-R\mathbf{c}
=
\boxed{
\begin{bmatrix}
3+4\sqrt3\\
4-3\sqrt3
\end{bmatrix}}
.$$

Hence the transformation is a $60^\circ$ rotation about the origin followed by translation by $\mathbf{p}$. Its homogeneous matrix is

$$
\begin{bmatrix}
\tfrac12 & -\tfrac{\sqrt3}{2} & 3+4\sqrt3\\
\tfrac{\sqrt3}{2} & \tfrac12 & 4-3\sqrt3\\
0 & 0 & 1
\end{bmatrix}.
$$

**Knowledge point and takeaway.** Rotation about an arbitrary point is a conjugated transformation $T_{\mathbf{c}}RT_{-\mathbf{c}}$. The formula $\mathbf{p}=\mathbf{c}-R\mathbf{c}$ is especially useful in CAD, robotics, and camera control.

# Section 2.7 - Exercise 16

Are $(1,-2,3,4)$ and $(10,-20,30,40)$ homogeneous coordinates for the same point in $\mathbb{R}^3$? Why or why not?

## Solution

Yes. Two nonzero homogeneous-coordinate vectors represent the same point when one is a nonzero scalar multiple of the other. Here,

$$
(10,-20,30,40)=10(1,-2,3,4).
$$

Equivalently, divide the first three entries by the fourth entry:

$$
\left(\frac14,-\frac12,\frac34\right)
=
\left(\frac{10}{40},\frac{-20}{40},\frac{30}{40}\right).
$$

Thus both vectors are homogeneous coordinates for

$$
\boxed{\left(\frac14,-\frac12,\frac34\right)}.
$$

**Knowledge point and takeaway.** Homogeneous coordinates are scale-invariant: multiplying all coordinates by the same nonzero number does not change the represented point. This freedom makes perspective transformations possible with matrix multiplication.

Let $S$ be the triangle with vertices $(4.2,1.2,4)$, $(6,4,2)$, $(2,2,6)$. Find the image of $S$ under the perspective projection with center of projection at $(0,0,10)$.

## Solution

For a perspective projection onto the $xy$-plane with center of projection $(0,0,d)$, Section 2.7 gives

$$
(x,y,z)\longmapsto
\left(\frac{x}{1-z/d},\frac{y}{1-z/d},0\right).
$$

Here $d=10$.

For $(4.2,1.2,4)$, the denominator is $1-4/10=0.6$, so the image is $(7,2,0)$.

For $(6,4,2)$, the denominator is $1-2/10=0.8$, so the image is $(7.5,5,0)$.

For $(2,2,6)$, the denominator is $1-6/10=0.4$, so the image is $(5,5,0)$.

Therefore, the projected triangle has vertices

$$
\boxed{(7,2,0),\quad(7.5,5,0),\quad(5,5,0)}.
$$

The following beginner-friendly Python code checks the arithmetic.

```python
import numpy as np 
vertices = np.array([[4.2, 1.2, 4.0], [6.0, 4.0, 2.0], [2.0, 2.0, 6.0]]) 
d = 10.0  
denominators = 1.0 - vertices[:, 2] / d  
projected_xy = vertices[:, :2] / denominators[:, None]  
projected_vertices = np.column_stack((projected_xy, np.zeros(3)))  
print(projected_vertices)  
```

# Section 2.7 - Exercise 21

Exercises 21 and 22 concern the way in which color is specified for display in computer graphics. A color on a computer screen is encoded by three numbers $(R,G,B)$ that list the amount of energy an electron gun must transmit to red, green, and blue phosphor dots on the computer screen. (A fourth number specifies the luminance or intensity of the color.)

The actual color a viewer sees on a screen is influenced by the specific type and amount of phosphors on the screen. So each computer screen manufacturer must convert between the $(R,G,B)$ data and an international CIE standard for color, which uses three primary colors, called $X$, $Y$, and $Z$. A typical conversion for short-persistence phosphors is

$$
\begin{bmatrix}
0.61 & 0.29 & 0.150 \\
0.35 & 0.59 & 0.063 \\
0.04 & 0.12 & 0.787
\end{bmatrix}
\begin{bmatrix}
R \\
G \\
B
\end{bmatrix}
=
\begin{bmatrix}
X \\
Y \\
Z
\end{bmatrix}.
$$

A computer program will send a stream of color information to the screen, using standard CIE data $(X,Y,Z)$. Find the equation that converts these data to the $(R,G,B)$ data needed for the screen's electron gun.

## Solution

Write the given equation as

$$
M\begin{bmatrix}R\\G\\B\end{bmatrix}
=
\begin{bmatrix}X\\Y\\Z\end{bmatrix}.
$$

To recover the screen data, multiply by $M^{-1}$:

$$
\begin{bmatrix}R\\G\\B\end{bmatrix}
=M^{-1}\begin{bmatrix}X\\Y\\Z\end{bmatrix}.
$$

Using the entries given in the exercise,

$$
\boxed{
\begin{bmatrix}R\\G\\B\end{bmatrix}
\approx
\begin{bmatrix}
2.2586 & -1.0395 & -0.3473\\
-1.3495 & 2.3441 & 0.0696\\
0.0910 & -0.3046 & 1.2777
\end{bmatrix}
\begin{bmatrix}X\\Y\\Z\end{bmatrix}}
.$$

The following Python code computes the inverse matrix directly.

```python
import numpy as np 
cie_matrix = np.array([[0.61, 0.29, 0.150], [0.35, 0.59, 0.063], [0.04, 0.12, 0.787]]) 
cie_to_rgb = np.linalg.inv(cie_matrix) 
print(np.round(cie_to_rgb, 4)) 
```

**Knowledge point and takeaway.** A change of color coordinates is a linear transformation, and reversing it requires an inverse matrix. In engineering, this is a device-calibration problem: the same standard CIE color must be converted into hardware-specific RGB commands.

## Solution

Let

The required conversion reverses this transformation:

Using the rounded entries supplied in the exercise,
