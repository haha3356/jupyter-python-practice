# Exercise 49

The matrix \(M\) can detect a particular $ 2 \times 2 $ colored pattern like in Example 9.

Create a nonzero $ 4 \times 1$ vector \(x\) by choosing each entry to be a zero or one.

Test to see if \(x\) corresponds to the right pattern by calculating:

$$
x^T M x
$$

If

$$
x^T M x = 0
$$

then \(x\) is the pattern identified by \(M\).

If

$$
x^T M x \neq 0
$$

try a different nonzero vector of zeros and ones.

The matrix is:

$$
M=
\begin{bmatrix}
1&0&-1&0\\
0&1&0&0\\
-1&0&1&0\\
0&0&0&1
\end{bmatrix}
$$

```python

import sympy as sp

M = sp.Matrix([
    [1, 0, -1, 0],
    [0, 1, 0, 0],
    [-1, 0, 1, 0],
    [0, 0, 0, 1]
])

for x1 in range(2):
    for x2 in range(2):
        for x3 in range(2):
            for x4 in range(2):

                
                x = sp.Matrix([
                    x1,
                    x2,
                    x3,
                    x4
                ])
                if x != sp.zeros(4, 1):
                    value = (x.T * M * x)[0]
                    if value == 0:
                        print(x)
```

Therefore, this vector represents the $ 2\times2 $ colored pattern detected by M.

Find out the number of digits your matrix program stores accurately, and report how many digits of accuracy are lost when \(x_1\) is used in place of the exact solution \(x\).

```python
import numpy as np  

A = np.array([[4, 0, -7, -7], 
              [-6, 1, 11, 9],  
              [7, -5, 10, 19], 
              [-1, 2, 3, -1]], dtype=float) 

condition_number = np.linalg.cond(A)  

print(condition_number)  

x=np.array([1,2,3,4],dtype=float)

b = A @ x  

x1 = np.linalg.solve(A, b) 

print(x)  

print(b)  

print(x1)  

error = np.abs(x - x1)

print(error)  
```

$log_{10}(23682.9) $ =4.37   
so the expected loss of accuracy is about:4 

Python uses double precision floating-point numbers, which normally store about:
15,16 significant digits.

The numerical experiment shows:
- x and x_1 agree to about 12 digits;
- the condition number predicts a loss of about 4 digits;
- the result is consistent with the theory of conditioning of linear systems.

# Exercise 53

Some matrix programs, such as MATLAB, have a command to create Hilbert matrices of various sizes.

If possible, use an inverse command to compute the inverse of a twelfth-order or larger Hilbert matrix:

$$
A^{-1}
$$

Compute:

$$
AA^{-1}
$$

and report what you find.



```python
import numpy as np  

n = 12  

A = np.zeros((n, n))  

for i in range(n):  

    for j in range(n): 

        A[i, j] = 1 / (i + j + 1)  

A_inverse = np.linalg.inv(A)  

result = A @ A_inverse 

print(result) 

condition_number = np.linalg.cond(A)  

print(condition_number)  
```

The experiment shows:
- A 12th-order Hilbert matrix is extremely ill-conditioned.
- Computing its inverse using standard floating-point arithmetic is unreliable.
- The product $ AA^{-1}$significantly deviates from the identity matrix.
- Therefore:
Large Hilbert matrices cannot be inverted accurately using ordinary floating-point arithmetic.
