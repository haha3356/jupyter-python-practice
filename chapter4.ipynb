## Section 4.1 - Vector Spaces and Subspaces

Main ideas: vector space axioms, the three subspace conditions, and spans.
A subset $H$ of a vector space $V$ is a subspace if it contains $\mathbf 0$
and is closed under addition and multiplication by every real scalar.
Theorem 1 says that a span of vectors in $V$ is a subspace of $V$.


# Section 4.1 - Exercise 5


In Exercises 5–8, determine if the given set is a subspace of $\mathbb P_n$ for
an appropriate value of $n$. Justify your answers.

All polynomials of the form $p(t)=at^2$, where $a$ is in $\mathbb R$.


## Solution


**Yes, it is a subspace of $\mathbb P_2$.** The set is

$$H=\{at^2:a\in\mathbb R\}=\operatorname{Span}\{t^2\}.$$

Since $t^2\in\mathbb P_2$, Theorem 1 applies. Directly, $a=0$ gives the zero
polynomial, and

$$at^2+bt^2=(a+b)t^2,\qquad c(at^2)=(ca)t^2$$

are again in $H$ for all real $a,b,c$.

# Section 4.1 - Exercise 8


In Exercises 5–8, determine if the given set is a subspace of $\mathbb P_n$ for
an appropriate value of $n$. Justify your answers.

All polynomials in $\mathbb P_n$ such that $p(0)=0$.


## Solution


**Yes.** Let $H=\{p\in\mathbb P_n:p(0)=0\}$. We use the subspace definition
from this section:

1. The zero polynomial satisfies $0(0)=0$.
2. If $p,q\in H$, then $(p+q)(0)=p(0)+q(0)=0$, so $p+q\in H$.
3. If $p\in H$ and $c\in\mathbb R$, then $(cp)(0)=cp(0)=0$, so $cp\in H$.

Thus $H$ is a subspace of $\mathbb P_n$. For $n\geq1$, it can also be written
as $\operatorname{Span}\{t,t^2,\ldots,t^n\}$. For $n=0$, it is the zero subspace.

# Section 4.1 - Exercise 20


The set of all continuous real-valued functions defined on a closed interval
$[a,b]$ in $\mathbb R$ is denoted by $C[a,b]$. This set is a subspace of the
vector space of all real-valued functions defined on $[a,b]$.

**(a)** What facts about continuous functions should be proved in order to
demonstrate that $C[a,b]$ is indeed a subspace as claimed? (These facts are
usually discussed in a calculus class.)

**(b)** Show that $\{f\text{ in }C[a,b]:f(a)=f(b)\}$ is a subspace of $C[a,b]$.


## Solution


**(a)** The three facts matching the subspace definition are:

1. The zero function is continuous on $[a,b]$.
2. The sum of two functions continuous on $[a,b]$ is continuous on $[a,b]$.
3. A real scalar multiple of a function continuous on $[a,b]$ is continuous
   on $[a,b]$.

**(b)** Write $H=\{f\in C[a,b]:f(a)=f(b)\}$. The zero function belongs to $H$.
If $f,g\in H$, then $f+g$ is continuous by part (a), and

$$
(f+g)(a)=f(a)+g(a)=f(b)+g(b)=(f+g)(b).
$$

So $f+g\in H$. If $c\in\mathbb R$, then $cf$ is continuous and

$$ (cf)(a)=cf(a)=cf(b)=(cf)(b), $$

so $cf\in H$. All three subspace conditions hold.

$\mathbf u+\mathbf w=\mathbf0$. Adding $-\mathbf u$ to both sides, we have

$$\begin{aligned}
(-\mathbf u)+[\mathbf u+\mathbf w]&=(-\mathbf u)+\mathbf0,\\
[(-\mathbf u)+\mathbf u]+\mathbf w&=(-\mathbf u)+\mathbf0
&&\text{by Axiom }\underline{\qquad}\text{ (a)},\\
\mathbf0+\mathbf w&=(-\mathbf u)+\mathbf0
&&\text{by Axiom }\underline{\qquad}\text{ (b)},\\
\mathbf w&=-\mathbf u&&\text{by Axiom }\underline{\qquad}\text{ (c)}.
\end{aligned}$$

- **Axiom 4 (additive identity)** simplifies both sides:
  $\mathbf0+\mathbf w=\mathbf w$ and $(-\mathbf u)+\mathbf0=-\mathbf u$.

Thus any additive inverse $\mathbf w$ equals $-\mathbf u$.

# Section 4.1 - Exercise 40


Let $H$ and $K$ be subspaces of a vector space $V$. The **intersection** of
$H$ and $K$, written as $H\cap K$, is the set of $\mathbf v$ in $V$ that
belong to both $H$ and $K$. Show that $H\cap K$ is a subspace of $V$.
(See the figure.) Give an example in $\mathbb R^2$ to show that the union of
two subspaces is not, in general, a subspace.




## Solution


Use the three subspace conditions:

1. Since $\mathbf0\in H$ and $\mathbf0\in K$, we have $\mathbf0\in H\cap K$.
2. If $\mathbf u,\mathbf v\in H\cap K$, then both vectors belong to both
   subspaces. Thus $\mathbf u+\mathbf v\in H$ and $\mathbf u+\mathbf v\in K$,
   so $\mathbf u+\mathbf v\in H\cap K$.
3. If $\mathbf u\in H\cap K$ and $c\in\mathbb R$, then $c\mathbf u\in H$ and
   $c\mathbf u\in K$, so $c\mathbf u\in H\cap K$.

Therefore $H\cap K$ is a subspace of $V$.

For the union, take the two coordinate axes in $\mathbb R^2$:

$$H=\left\{\begin{bmatrix}a\\0\end{bmatrix}:a\in\mathbb R\right\},
\qquad K=\left\{\begin{bmatrix}0\\b\end{bmatrix}:b\in\mathbb R\right\}.$$

Both are subspaces, but $(1,0)^T$ and $(0,1)^T$ belong to $H\cup K$, while
their sum $(1,1)^T$ belongs to neither axis. Thus $H\cup K$ is not closed
under addition and is not a subspace.


# Section 4.1 - Exercise 41


Given subspaces $H$ and $K$ of a vector space $V$, the **sum** of $H$ and $K$,
written as $H+K$, is the set of all vectors in $V$ that can be written as the
sum of two vectors, one in $H$ and the other in $K$; that is,

$$H+K=\{\mathbf w:\mathbf w=\mathbf u+\mathbf v\text{ for some }\mathbf u\in H
\text{ and some }\mathbf v\in K\}.$$

**(a)** Show that $H+K$ is a subspace of $V$.

**(b)** Show that $H$ is a subspace of $H+K$ and $K$ is a subspace of $H+K$.


## Solution


**(a)** First, $\mathbf0=\mathbf0+\mathbf0\in H+K$. If
$\mathbf w_1=\mathbf h_1+\mathbf k_1$ and
$\mathbf w_2=\mathbf h_2+\mathbf k_2$, where $\mathbf h_i\in H$ and
$\mathbf k_i\in K$, then

$$\mathbf w_1+\mathbf w_2=(\mathbf h_1+\mathbf h_2)+(\mathbf k_1+\mathbf k_2)\in H+K,$$

because $H$ and $K$ are closed under addition. For any real $c$,

$$c\mathbf w_1=c\mathbf h_1+c\mathbf k_1\in H+K,$$

because each subspace is closed under scalar multiplication. Thus $H+K$ is
a subspace of $V$.

**(b)** Every $\mathbf h\in H$ can be written as $\mathbf h+\mathbf0$ with
$\mathbf0\in K$, so $H\subseteq H+K$. Similarly $K\subseteq H+K$.
The zero vector, sums, and scalar multiples in $H$ and $K$ are the same as
in $V$, so their existing subspace properties also make them subspaces of
$H+K$.


# Section 4.1 - Exercise 42


Suppose $\mathbf u_1,\ldots,\mathbf u_p$ and $\mathbf v_1,\ldots,\mathbf v_q$
are vectors in a vector space $V$, and let

$$H=\operatorname{Span}\{\mathbf u_1,\ldots,\mathbf u_p\}
\quad\text{and}\quad
K=\operatorname{Span}\{\mathbf v_1,\ldots,\mathbf v_q\}.$$

Show that
$H+K=\operatorname{Span}\{\mathbf u_1,\ldots,\mathbf u_p,\mathbf v_1,\ldots,\mathbf v_q\}$.


## Solution


We prove both inclusions using the definition of span.

If $\mathbf w\in H+K$, then $\mathbf w=\mathbf h+\mathbf k$ with
$\mathbf h\in H$ and $\mathbf k\in K$. Thus there are scalars $a_i,b_j$ such that

$$\mathbf w=\sum_{i=1}^p a_i\mathbf u_i+\sum_{j=1}^q b_j\mathbf v_j.$$

This is a linear combination of the combined list, so $\mathbf w$ belongs
to its span.

Conversely, any vector in that span has the displayed form. Its first sum
belongs to $H$ and its second sum belongs to $K$, so it belongs to $H+K$.
The two sets are therefore equal.


# Section 4.1 - Exercise 44


Determine if $\mathbf y$ is in the subspace of $\mathbb R^4$ spanned by the
columns of $A$, where

$$\mathbf y=\begin{bmatrix}-4\\-8\\6\\-5\end{bmatrix},\qquad
A=\begin{bmatrix}3&-5&-9\\8&7&-6\\-5&-8&3\\2&-2&-9\end{bmatrix}.$$


## Solution


By the definition of span, we need scalars $x_1,x_2,x_3$ satisfying
$A\mathbf x=\mathbf y$. Row reduction gives

$$\left[\begin{array}{rrr|r}
3&-5&-9&-4\\8&7&-6&-8\\-5&-8&3&6\\2&-2&-9&-5
\end{array}\right]
\sim
\left[\begin{array}{ccc|r}
1&0&0&-1/5\\0&1&0&-2/5\\0&0&1&3/5\\0&0&0&0
\end{array}\right].$$

The system is consistent. If $\mathbf a_1,\mathbf a_2,\mathbf a_3$ are the
columns of $A$, then

$$\boxed{\mathbf y=-\tfrac15\mathbf a_1-\tfrac25\mathbf a_2+\tfrac35\mathbf a_3.}$$

Therefore **yes**, $\mathbf y$ is in the required subspace. The code performs
this same row reduction and checks the resulting linear combination.


```python

import sympy as sp

from IPython.display import display

A = sp.Matrix([
    [3, -5, -9],
    [8, 7, -6],
    [-5, -8, 3],
    [2, -2, -9]
])

y = sp.Matrix([-4, -8, 6, -5])


augmented = A.row_join(y)

reduced, pivots = augmented.rref()
print('Reduced augmented matrix:')
display(reduced)
```

# Section 4.1 - Exercise 45


The vector space $H=\operatorname{Span}\{1,\cos^2t,\cos^4t,\cos^6t\}$ contains
at least two interesting functions that will be used in a later exercise:

$$f(t)=1-8\cos^2t+8\cos^4t,$$
$$g(t)=-1+18\cos^2t-48\cos^4t+32\cos^6t.$$

Study the graph of $f$ for $0\leq t\leq2\pi$, and guess a simple formula for
$f(t)$. Verify your conjecture by graphing the difference between $1+f(t)$
and your formula for $f(t)$. (Hopefully, you will see the constant function
$1$.) Repeat for $g$.


## Solution


The graph of $f$ suggests $\cos4t$ (four cycles on $[0,2\pi]$), and the graph
of $g$ suggests $\cos6t$ (six cycles). Thus

$$\boxed{f(t)=\cos4t,\qquad g(t)=\cos6t.}$$

The requested difference graphs are

$$1+f(t)-\cos4t=1,\qquad 1+g(t)-\cos6t=1.$$

The code plots the original functions, their conjectured formulas, and these
differences. Tiny errors from floating-point arithmetic are expected.

For an algebraic verification, put $u=\cos^2t$ and use the double-angle formula:

$$\cos4t=2(\cos2t)^2-1=2(2u-1)^2-1=1-8u+8u^2.$$

Also, with $z=\cos2t=2u-1$, the triple-angle formula gives

$$\cos6t=4z^3-3z=4(2u-1)^3-3(2u-1)
=-1+18u-48u^2+32u^3.$$

These identities explain why the simpler functions belong to the span $H$.


```python

import numpy as np

import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 1000)

f = 1 - 8 * np.cos(t)**2 + 8 * np.cos(t)**4
g = -1 + 18 * np.cos(t)**2 - 48 * np.cos(t)**4 + 32 * np.cos(t)**6

for name, values, frequency in [('f', f, 4), ('g', g, 6)]:

    guess = np.cos(frequency * t)

    difference = 1 + values - guess

    fig, axes = plt.subplots(1, 2, figsize=(11, 3))

    axes[0].plot(t, values, label=name + '(t)')

    axes[0].plot(t, guess, '--', label=f'cos({frequency}t)')
    axes[0].set_title('Original function and conjecture')

    axes[0].legend()
    axes[1].plot(t, difference)
  
    axes[1].axhline(1, color='black', linestyle='--', label='y = 1')

    axes[1].set_ylim(0.9, 1.1)
    axes[1].set_title(f'1 + {name}(t) - cos({frequency}t)')
    axes[1].legend()
  
    for ax in axes:
        ax.set_xlim(0, 2 * np.pi)
        ax.set_xlabel('t (radians)')
        ax.set_ylabel('Function value')

        ax.grid(True, alpha=0.3)

    plt.tight_layout()

    plt.show()

    print(name, ': maximum absolute difference from the conjecture =',
          np.max(np.abs(values - guess)))
```

## Solution


Every vector $A\mathbf x$ has $m$ entries, so $\operatorname{Col}A\subseteq\mathbb R^m$.

## Solution


For continuous $f$, the Fundamental Theorem of Calculus gives the unique

$$T(f)(t)=\int_0^t f(s)\,ds.$$

$$T(f+g)(t)=\int_0^t[f(s)+g(s)]\,ds=T(f)(t)+T(g)(t),$$

where $0$ denotes the zero function. Nonzero constant functions are not

Exercises 41 and 42 reveal an important connection between linear
independence and linear transformations and provide practice using the
definition of linear dependence. Let $V$ and $W$ be vector spaces, let
$T:V\to W$ be a linear transformation, and let
$\{\mathbf v_1,\ldots,\mathbf v_p\}$ be a subset of $V$.


Show that if $\{\mathbf v_1,\ldots,\mathbf v_p\}$ is linearly dependent in
$V$, then the set of images, $\{T(\mathbf v_1),\ldots,T(\mathbf v_p)\}$,
is linearly dependent in $W$. This fact shows that if a linear transformation
maps a set $\{\mathbf v_1,\ldots,\mathbf v_p\}$ onto a linearly independent
set $\{T(\mathbf v_1),\ldots,T(\mathbf v_p)\}$, then the original set is
linearly independent, too (because it cannot be linearly dependent).


## Solution

Since the original vectors are linearly dependent, there are scalars
$c_1,\ldots,c_p$, **not all zero**, such that

$$c_1\mathbf v_1+\cdots+c_p\mathbf v_p=\mathbf0_V.$$

Apply $T$ and use linearity:

$$
c_1T(\mathbf v_1)+\cdots+c_pT(\mathbf v_p)
=T(c_1\mathbf v_1+\cdots+c_p\mathbf v_p)
=T(\mathbf0_V)=\mathbf0_W.
$$

The coefficients have not changed, so they are still not all zero. This
is a nontrivial dependence relation among the images. Hence the images
are linearly dependent.

Consequently, independent images force the original vectors to be
independent: dependence of the original vectors would give the contradiction
just proved.

```python
import sympy as sp
from IPython.display import display

sample_times = [0, sp.pi, sp.pi / 2, sp.pi / 4]

rows = []

for t in sample_times:


    row = [t, sp.sin(t), sp.cos(2 * t), sp.sin(t) * sp.cos(t)]
    rows.append(row)

A = sp.Matrix(rows)
print('Coefficient matrix of the four homogeneous equations:')
display(A)

reduced, pivots = A.rref()
print('Reduced matrix (I4 means every coefficient must be zero):')
display(reduced)
```

# Section 4.3 - Exercise 48

Show that $\{1,\cos t,\cos^2t,\ldots,\cos^6t\}$ is a linearly independent
set of functions defined on $\mathbb R$. Use the method of Exercise 47.
(This result will be needed in Exercise 54 in Section 4.5.)


## Solution

Suppose that

$$c_0+c_1\cos t+c_2\cos^2t+\cdots+c_6\cos^6t=0
\qquad\text{for every real }t.$$

Following Exercise 47, evaluate this identity at seven points. Choose
$t_j=\arccos x_j$, where the seven cosine values are

$$x_j\in\left\{-1,-\tfrac23,-\tfrac13,0,\tfrac13,\tfrac23,1\right\}.$$

Every value is in $[-1,1]$, so each sample time is real. At each sample
the equation is

$$c_0+c_1x_j+c_2x_j^2+\cdots+c_6x_j^6=0.$$

The resulting coefficient matrix has row $j$ equal to
$[\,1\ x_j\ x_j^2\ \cdots\ x_j^6\,]$. Exact row reduction in the code below
gives $I_7$, so every $c_j$ is zero. Thus the seven functions are linearly
independent.

This is a sufficient test: any relation valid for **all** real $t$ must
satisfy these seven equations, and those equations admit only the zero
coefficient vector. All fractions are kept exact.


```python
import sympy as sp
from IPython.display import display


cosine_values = [
    -1, -sp.Rational(2, 3), -sp.Rational(1, 3), 0,
    sp.Rational(1, 3), sp.Rational(2, 3), 1
]

rows = []

for x in cosine_values:

    row = []

    for power in range(7):
        row.append(x ** power)
    rows.append(row)
    
A = sp.Matrix(rows)
reduced, pivots = A.rref()
print('Reduced coefficient matrix:')
display(reduced)
print('Number of pivot columns:', len(pivots))

print('All seven coefficients must be zero:', reduced == sp.eye(7))
```

Exercises 27–30 concern a vector space $V$, a basis
$\mathcal B=\{\mathbf b_1,\ldots,\mathbf b_n\}$, and the coordinate mapping
$\mathbf x\mapsto[\mathbf x]_{\mathcal B}$.


Show that the coordinate mapping is one-to-one.
[Hint: Suppose $[\mathbf u]_{\mathcal B}=[\mathbf w]_{\mathcal B}$ for some

## Solution

Suppose that $[\mathbf u]_{\mathcal B}=[\mathbf w]_{\mathcal B}$.
Call their common entries $c_1,\ldots,c_n$. By the definition of coordinates,

$$\mathbf u=c_1\mathbf b_1+\cdots+c_n\mathbf b_n,
\qquad
\mathbf w=c_1\mathbf b_1+\cdots+c_n\mathbf b_n.$$

The right sides are identical, so $\mathbf u=\mathbf w$. Therefore equal
coordinate vectors come from equal vectors, which proves that the
coordinate mapping is one-to-one.


# Section 4.4 - Exercise 28

Exercises 27–30 concern a vector space $V$, a basis
$\mathcal B=\{\mathbf b_1,\ldots,\mathbf b_n\}$, and the coordinate mapping
$\mathbf x\mapsto[\mathbf x]_{\mathcal B}$.


Show that the coordinate mapping is onto $\mathbb R^n$. That is, given any
$\mathbf y$ in $\mathbb R^n$, with entries $y_1,\ldots,y_n$, produce
$\mathbf u$ in $V$ such that $[\mathbf u]_{\mathcal B}=\mathbf y$.


## Solution

Given any $\mathbf y=(y_1,\ldots,y_n)^T\in\mathbb R^n$, define

$$\boxed{\mathbf u=y_1\mathbf b_1+\cdots+y_n\mathbf b_n.}$$

This is a vector in $V$ because $V$ is closed under linear combinations.
Its coefficients relative to the basis $\mathcal B$ are $y_1,\ldots,y_n$.
The Unique Representation Theorem guarantees that these are its coordinates,
so

$$[\mathbf u]_{\mathcal B}=\begin{bmatrix}y_1\\\vdots\\y_n\end{bmatrix}=\mathbf y.$$

Every vector in $\mathbb R^n$ is reached, proving that the coordinate mapping
is onto.


# Section 4.4 - Exercise 29

Exercises 27–30 concern a vector space $V$, a basis
$\mathcal B=\{\mathbf b_1,\ldots,\mathbf b_n\}$, and the coordinate mapping
$\mathbf x\mapsto[\mathbf x]_{\mathcal B}$.


Show that a subset $\{\mathbf u_1,\ldots,\mathbf u_p\}$ in $V$ is linearly
independent if and only if the set of coordinate vectors
$\{[\mathbf u_1]_{\mathcal B},\ldots,[\mathbf u_p]_{\mathcal B}\}$ is linearly
independent in $\mathbb R^n$. [Hint: Since the coordinate mapping is
one-to-one, the following equations have the same solutions $c_1,\ldots,c_p$.]

$$c_1\mathbf u_1+\cdots+c_p\mathbf u_p=\mathbf0_V,$$
$$[c_1\mathbf u_1+\cdots+c_p\mathbf u_p]_{\mathcal B}
=[\mathbf0_V]_{\mathcal B}=\mathbf0_{\mathbb R^n}.$$


## Solution

The coordinate mapping is linear, and it sends $\mathbf0_V$ to the zero
coordinate vector. Since it is also one-to-one, for any coefficients $c_j$,

$$\begin{aligned}
c_1\mathbf u_1+\cdots+c_p\mathbf u_p=\mathbf0_V
&\iff [c_1\mathbf u_1+\cdots+c_p\mathbf u_p]_{\mathcal B}=\mathbf0\\
&\iff c_1[\mathbf u_1]_{\mathcal B}+\cdots
+c_p[\mathbf u_p]_{\mathcal B}=\mathbf0.
\end{aligned}$$

Thus the two homogeneous equations have exactly the same coefficient
solutions. The original vectors have only the trivial relation if and only
if their coordinate vectors have only the trivial relation. This is the
required equivalence of linear independence.

Therefore the linear-combination condition is preserved in both directions,
with the **same coefficients**.

## Section 4.5 - The Dimension of a Vector Space



# Section 4.5 - Exercise 51

Exercises 51 and 52 concern finite-dimensional vector spaces $V$ and $W$
and a linear transformation $T:V\to W$.


Let $H$ be a nonzero subspace of $V$, and let $T(H)$ be the set of images
of vectors in $H$. Then $T(H)$ is a subspace of $W$, by Exercise 47 in
Section 4.2. Prove that $\dim T(H)\leq\dim H$.


## Solution

Let $\{\mathbf h_1,\ldots,\mathbf h_r\}$ be a basis for $H$, so $\dim H=r$.
For any $\mathbf h\in H$, there are scalars $c_1,\ldots,c_r$ such that

$$\mathbf h=c_1\mathbf h_1+\cdots+c_r\mathbf h_r.$$

Applying the linear transformation gives

$$T(\mathbf h)=c_1T(\mathbf h_1)+\cdots+c_rT(\mathbf h_r).$$

Thus every vector in $T(H)$ is in the span of the $r$ images. Conversely,
any linear combination of those images equals
$T(c_1\mathbf h_1+\cdots+c_r\mathbf h_r)$ and therefore belongs to $T(H)$.
Hence

$$T(H)=\operatorname{Span}\{T(\mathbf h_1),\ldots,T(\mathbf h_r)\}.$$

If $T(H)=\{\mathbf0\}$, its dimension is $0\leq r$. Otherwise, by the
Spanning Set Theorem, a basis for $T(H)$ can be selected from these at most
$r$ images. Consequently,

$$\boxed{\dim T(H)\leq r=\dim H.}$$

The images need not be independent: some may be zero or may be combinations
of the others. This is why the dimension can decrease.


# Section 4.5 - Exercise 52

Exercises 51 and 52 concern finite-dimensional vector spaces $V$ and $W$
and a linear transformation $T:V\to W$.


Let $H$ be a nonzero subspace of $V$, and suppose $T$ is a one-to-one (linear)
mapping of $V$ into $W$. Prove that $\dim T(H)=\dim H$. If $T$ happens to be
a one-to-one mapping of $V$ onto $W$, then $\dim V=\dim W$. Isomorphic
finite-dimensional vector spaces have the same dimension.


## Solution

Let $\{\mathbf h_1,\ldots,\mathbf h_r\}$ be a basis for $H$. Exercise 51
shows that the list $T(\mathbf h_1),\ldots,T(\mathbf h_r)$ spans $T(H)$.
We show that this list is also linearly independent.

Suppose

$$c_1T(\mathbf h_1)+\cdots+c_rT(\mathbf h_r)=\mathbf0_W.$$

Linearity gives

$$T(c_1\mathbf h_1+\cdots+c_r\mathbf h_r)=\mathbf0_W=T(\mathbf0_V).$$

Because $T$ is one-to-one, equal outputs have equal inputs. Thus

$$c_1\mathbf h_1+\cdots+c_r\mathbf h_r=\mathbf0_V.$$

The $\mathbf h_j$ form a basis and are independent, so
$c_1=\cdots=c_r=0$. Therefore the $r$ images form a basis for $T(H)$, and

$$\boxed{\dim T(H)=r=\dim H.}$$

If $T$ is also onto $W$, take $H=V$. Then $T(V)=W$, so the same result gives

$$\boxed{\dim W=\dim T(V)=\dim V.}$$

According to Theorem 12, a linearly independent set
