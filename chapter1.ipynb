# Section 1.2 - Exercise 46

In a wind tunnel experiment, the force on a projectile due to air resistance was measured at different velocities:

| Velocity (100 ft/sec) | 0 | 2 | 4 | 6 | 8 | 10 |
|-----------------------|---|---|---|---|---|----|
| Force (100 lb)        | 0 | 2.90 | 14.8 | 39.6 | 74.3 | 119 |

Find an interpolating polynomial for these data and estimate the force on the projectile when the projectile is traveling at **750 ft/sec**.

Use:

\[
p(t)=a_0+a_1t+a_2t^2+a_3t^3+a_4t^4+a_5t^5
\]

What happens if you try to use a polynomial of degree less than 5?  
(Try a cubic polynomial, for instance.)

## Solution

```python
import sympy as sp
A=sp.Matrix([[1, 0, 0, 0, 0, 0],
             [1, 2, 4, 8, 16, 32],
             [1, 4, 16, 64, 256, 1024],
             [1, 6, 36, 216, 1296, 7776],
             [1, 8, 64, 512, 4096, 32768],
             [1, 10, 100, 1000, 10000, 100000]])
b=sp.Matrix([[0, 2.90, 14.8, 39.6, 74.3, 119]])
ans=sp.linsolve((A,b))
print(ans)
```

```python
import numpy as np  

V= np.array([0, 2, 4, 6, 8, 10])  

F= np.array([0, 2.90, 14.8, 39.6, 74.3, 119])  

P5 = np.polyfit(V, F, 5)  

print(P5)  

t = 7.5  

f_1 = np.polyval(P5, t)  

print(f_1)  

F_1= f_1 * 100  

print(F_1)  
```

The interpolation polynomial is:

$$
P(t)=0+1.7125t-1.1948t^2+0.6615t^3-0.0701t^4+0.0026t^5
$$

P(7.5)$\approx$6484(1b)

```python
import numpy as np  

V= np.array([0, 2, 4, 6, 8, 10])  

F= np.array([0, 2.90, 14.8, 39.6, 74.3, 119])  

P3 = np.polyfit(V, F, 3)  

print(P3)  

f_3 = np.polyval(P3, t)  

print(f_3)  

F_3= f_3 * 100  

print(F_3)  
```

A cubic least-squares fit gives approximately:

\[

\]

has the form

\[

A linear transformation must satisfy:T(0)=0

Now evaluate the T(0):
T(0)=A*0+b=b

When:

$
T(0)=b\neq0

A car rented at one location may be returned to any of the three locations. The fractions of cars returned to the three locations are shown in the matrix below.

The columns represent the location where cars are rented from, and the rows represent the location where cars are returned to.

\[
A=
\begin{bmatrix}
0.97 & 0.05 & 0.10\\
0.00 & 0.90 & 0.05\\
0.03 & 0.05 & 0.85
\end{bmatrix}
\]

where:

- Row 1: Airport returns
- Row 2: East side office returns
- Row 3: West side office returns

and

- Column 1: Cars rented from Airport
- Column 2: Cars rented from East office
- Column 3: Cars rented from West office

---

Suppose that on Monday there are:

\[
x_0=
\begin{bmatrix}
295\\
55\\
150
\end{bmatrix}
\]

cars at:

- Airport: 295 cars
- East side office: 55 cars
- West side office: 150 cars

Find the approximate distribution of cars on Wednesday.

Use the transition matrix:

\[
x_{n+1}=Ax_n
\]

Since Wednesday is two days after Monday:

\[
x_2=A^2x_0
\]

Calculate:

x_1=Ax_0
\]

then

x_2=Ax_1
\]
