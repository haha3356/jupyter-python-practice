```python
import numpy as np  

V= np.array([0, 2, 4, 6, 8, 10])  

F= np.array([0, 2.90, 14.8, 39.6, 74.3, 119])  

P5 = np.polyfit(V, F, 5)  

print(P5)  

t = 7.5  

f_1 = np.polyval(P5, t)  

print(f_1)  

F_1= f_1 * 100  

print(F_1)  
```

The interpolation polynomial is:

$$
P(t)=0+1.7125t-1.1948t^2+0.6615t^3-0.0701t^4+0.0026t^5
$$

P(7.5)$\approx$6484(1b)

```python
import numpy as np  

V= np.array([0, 2, 4, 6, 8, 10])  

F= np.array([0, 2.90, 14.8, 39.6, 74.3, 119])  

P3 = np.polyfit(V, F, 3)  

print(P3)  

f_3 = np.polyval(P3, t)  

print(f_3)  

F_3= f_3 * 100  

print(F_3)  
```

A cubic least-squares fit gives approximately:

$$
P3(t)=0.1754-1.7412t+1.4096t^2-0.0046t^3
$$

P3(7.5)$\approx$6445(1b)

A cubic polynomial cannot interpolate all six points simultaneously.

```python
import sympy as sp

A45 = sp.Matrix([
    [-5,10,-5,4],
    [8,3,-4,7],
    [4,-9,5,-3],
    [-3,-2,5,4]
])

A46 = sp.Matrix([
    [7,5,4,-9],
    [10,6,16,-4],
    [12,8,12,7],
    [-8,-6,-2,5]
])

A47 = sp.Matrix([
    [4,-7,3,7,5],
    [6,-8,5,12,-8],
    [-7,10,-8,-9,14],
    [3,-5,4,2,-6],
    [-5,6,-6,-7,3]
])

A48 = sp.Matrix([
    [9,13,5,6,-1],
    [14,15,-7,-6,4],
    [-8,-9,12,-5,-9],
    [-5,-6,-8,9,8],
    [13,14,15,2,11]
])

A45_rref=A45.rref()[0]
A46_rref=A46.rref()[0]
A47_rref=A47.rref()[0]
A48_rref=A48.rref()[0]

print(A45_rref)
print(A46_rref)
print(A47_rref)
print(A48_rref)
```

There are 4 columns, but only 3 pivots.

Therefore:A45 is not one-to-one

For A46_rref:
There are 4 columns, but only 3 pivots.

Therefore:A46 is not one-to-one

For A47_rref:
There are 5 rows, but only 4 pivots.

Therefore:A47 is not onto $R^5$

For A48_rref:
There are 5 rows, but only 4 pivots.

Therefore:A48 is not onto $R^5$
